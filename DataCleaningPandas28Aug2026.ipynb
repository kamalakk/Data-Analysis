```python
import pandas as pd
#pd.set_option('display.max.columns', None)
#pd.set_option('display.max.rows', None)
df=pd.read_csv('/content/world_population.csv')
df.head(5)
```

```python
#df=df.drop(columns=["CCA3"])
df.head(5)
```

```python
df=df.drop(columns=['1990 Population','1980 Population'])
df.head(5)
```

```python
df=df.dropna(subset=['2022 Population'])
df.head(5)
```

```python
print(df.to_string())
```

```python
df=df.fillna({"Country": "None"})
df.head(5)
```

```python
#df['Continent']=df['Continent'].replace(to_replace="Oceania", value="New Continent")
df['Capital']=df['Capital'].replace({"Kabul":"Afghan Capital","Tirana":"Albabia Capital"})
df.head(5)
```

```python
df['Country']=df['Country'].str.lower()
df.head(5)
```

```python
#df['columnname']=df['columnname'].astype(bool)
df=df.drop_duplicates()
```
