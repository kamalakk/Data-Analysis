Exploratory Data Analysis in Pandas

```python
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt

df=pd.read_csv('/world_population.csv')
#pd.set_option('display.max_rows',None)
df.head(5)

```

```python
pd.set_option('display.float_format', lambda x: '%.3f' % x)
df
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.nunique()
```

```python
df.sort_values(by='2022 Population', ascending=False).head()
```

```python
df.select_dtypes(include=['number']).corr()
```

```python
sb.heatmap(df.select_dtypes(include=['number']).corr(), annot=True, fmt = '.2f')
plt.rcParams['figure.figsize']=(10,10)
plt.show()
```

```python
df2=df.groupby('Continent').mean(numeric_only=True).sort_values(by='2022 Population', ascending=False)
```

```python
df[df['Continent'].str.contains('Oceania')]
```

```python
df3=df2.transpose()
df3
```

```python

df3.plot()
```
