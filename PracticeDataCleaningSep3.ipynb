```python
import pandas as pd
import numpy as np
from word2number import w2n
# A deliberately messy dataset
data = {
    'Name': [' John Doe ', 'JANE smith', 'Bob Johnson', ' john doe', 'ALICE WONG',
             'Bob Johnson', 'Mike Chen', 'Sara Lee', None, 'jane smith'],
    'Age': [28, '32', 45, 28, 'twenty-nine', 45, 37, 26, 40, '32'],
    'Email': ['john@x.com', 'jane@x.com', 'bob@X.com', 'JD@x.com', 'alice@x.com',
              'bob@x.com', None, 'sara@x.com', 'mr_x@x.com', 'JANE@x.com'],
    'Salary': ['$45,000', '$52,000', '$78,500', '$45,000', 'error',
               '$78,500', '$61,200', '$48,900', 'N/A', '$52,000'],
    'Join_Date': ['2021-05-01', '05/06/2020', '2022-01-15', '2021-05-01',
                  '2020-11-30', '15/01/2022', None, '2021-08-20', '2019-03-04', '2020-06-05'],
    'City': ['NYC', 'Los Angeles', 'Chicago', 'nyc', 'SFO', 'Chicago',
             'Austin', 'Seattle', 'NYC', 'LA']
}
```

Age column

Convert to numeric, handling the text value 'twenty-nine' (map it to 29)
Report how many null/missing result

```python
df.info()
```

```python
def convert_age(x):
    try:
        # Try converting as integer first
        return int(x)
    except ValueError:
        try:
            # Convert word to number
            return w2n.word_to_num(x)
        except:
            # If both fail, return NaN or some default value
            return None
# Apply the function to the 'age' column
df['Age'] = df['Age'].apply(convert_age)
```

```python
df['Salary'] = df['Salary'].replace(['NA', 'Missing', 'missing', 'N/A','error', ''], np.nan)

df['Salary'] = (
    df['Salary']
    .str.replace('[\$,]', '', regex=True)
    .astype(float)          # intermediate float allows NaN to survive
    .astype('Int64')        # nullable integer dtype (keeps NaN)
)
df
```

```python
df['Join_Date']=df['Join_Date'].str.replace('/', '-')
df
```

```python
df['Join_Date'] = pd.to_datetime(df['Join_Date'], format='mixed')

# 2. Convert to the specific dd-mmm-yyyy string format
df['Join_Date'] = df['Join_Date'].dt.strftime('%d-%b-%Y')
```

```python
city_disc = {
    'NYC': 'New York',
    'Los Angeles': 'Los Angeles',
    'SFO': 'San Francisco',
    'Chicago': 'Chicago',
    'Seattle': 'Seattle',
    'nyc' : 'New York'
}
```

```python
df['City'] = df['City'].str.upper()
city_mapping_upper_keys = {k.upper(): v for k, v in city_disc.items()}
df['City'] = df['City'].replace(city_mapping_upper_keys)
df['City'] = df['City'].str.title()
df
```
