Group by pandas 2nd september 2026

```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
sns.get_dataset_names()
```

```python
df=sns.load_dataset('tips')
df.head()
```

```python
df.groupby('time',observed=True)['total_bill'].mean()
```

```python
df.groupby('time',observed=True)['total_bill'].mean().to_frame('Avg-Bill')
```

```python
df.groupby('time',observed=True)['total_bill'].mean().to_frame('Avg-Bill').reset_index()
```

```python
df.groupby('time',observed=True)['total_bill'].mean().to_frame('Avg-Bill').reset_index().sort_values('Avg-Bill', ascending=False)
```

```python
avgbill=df.groupby('time',observed=True)['total_bill'].mean().to_frame('Avg-Bill').reset_index().sort_values('Avg-Bill', ascending=False)
avgbill.head()
```

```python
df.groupby('time', observed=True)['total_bill'].count()
```

```python
df.groupby('time', observed=True)['total_bill'].size()
```

```python
df.groupby('time', observed=True)['total_bill'].nunique()
```

```python
df.groupby('time', observed=True)['total_bill'].max()
```

```python
df.groupby('time', observed=True)['total_bill'].min()
```

```python
df.groupby('time', observed=True)['total_bill'].sum()
```
