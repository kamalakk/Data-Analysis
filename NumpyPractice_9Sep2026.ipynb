```python
import numpy as np
```

```python
from re import M
months= np.array(['January','February','March','April','May','June','July','August','September','October','November','December'])
sales=[]

for x in months:
     value = float(input ("Enter {x} sales value : " ))
     sales.append(value)

sales=np.array(sales)
#print(months)
#print(sales)

print("---------Total Sales--------")
print("Total Sales  :", np.sum(sales))
print("Average Sales:", np.mean(sales))
print("Max Sales    :", np.max(sales))
print("Min Sales    :", np.min(sales))

best_month = months[np.argmax(sales)]
worst_onth = months[np.argmin(sales)]

print("Best Month of the Year is  :", best_month)
print("Worst Month of the Year is  :",worst_onth)


above_avg = months[sales > np.mean(sales)]
below_avg = months[sales < np.mean(sales)]

print("Months with above average sales :", above_avg)
print("Months with below average sales :", below_avg)
```
