Data Cleaning using python

```python
import pandas as pd
df= pd.read_csv('/hotel_bookings.csv')
df
```

```python
df.columns
```

```python
#df.drop(columns=['hotel'],inplace=True)
pd.set_option('display.max_columns', None)
#pd.set_option(display.str.columns, None)
df.rename({'adults':'num_adults', 'children':'num_children','babies':'num_babies'}, axis=1, inplace=True)
df.head(5)
```

```python
df.isnull().sum()*100/len(df)
```

```python
df['agent'].unique()
```

```python
df[df['agent'] == 5]
```

```python
df.fillna({'agent': -1}, inplace = True)
df
```

```python
df.columns
```

```python
df[df['num_children'].isna()]
```

```python
df[df['country'].isna()][['lead_time','arrival_date_year','arrival_date_month']]
```

```python
df['country'].fillna('Unknown', inplace = True)
df
```

```python
df=df.dropna(subset=['is_canceled'])
df
```

```python
df.isnull().sum() * 100 / len(df)
```
